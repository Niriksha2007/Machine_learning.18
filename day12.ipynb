```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Age': [20, 25, 28, 30, 32, 35, 40, 47, 50, 21, 38, 44, 33, 49, 33, 43],
    'Salary': [20000, 30000, 40000, 50000, 60000, 72000, 34000, 76000, 28000, 66000, 45000, 29000, 44000, 72000, 89000, 90000]
}

df = pd.DataFrame(data)
df
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()    #here we have created an object of MinMaxScaler class. This class is used to scale the data between 0 and 1.

val = scaler.fit_transform(df)
val
```

```python
np.round(val,2)  #here we are getting round off values upto 2 decimal places.
```

```python
dff=pd.DataFrame(val)
dff
```

```python
dff=pd.DataFrame(val,columns=['Age','Salary'])
dff
```

```python
from sklearn.cluster import KMeans
```

```python
km = KMeans(n_clusters=3, random_state=40)
km.fit_predict(dff)
```

```python
km.inertia_ #here it is printing SSE(Sum of square errors) value.
```

```python
sse_data=[]
for i in range(1,11):
    km = KMeans(n_clusters=i, random_state=40)
    km.fit_predict(dff)
    sse_data.append(km.inertia_)
sse_data
```

```python
np.round(sse_data, 4)
```
